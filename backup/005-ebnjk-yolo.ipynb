```python
MODEL_PATH = "/kaggle/input/yolo12-50/pytorch/default/1/yolo12-50.pt"  # Path to your trained .pt file
# DATASET_ROOT = "BirdNest-4"                         # Path to your dataset folder (e.g., "BirdNest-4")
DATASET_ROOT = "BirdNest-Balanced-1"                         # Path to your dataset folder (e.g., "BirdNest-4")

DATA_YAML_PATH = os.path.join(DATASET_ROOT, "data.yaml")
TEST_IMG_DIR = os.path.join(DATASET_ROOT, "test", "images")
TEST_LABEL_DIR = os.path.join(DATASET_ROOT, "test", "labels")

NUM_IMAGES = 64
CONF_THRESHOLD = 0.25  # Only show predictions above this confidence
GT_COLOR = (0, 255, 0)  # Green for Ground Truth
PRED_COLOR = (0, 0, 255) # Red for Prediction
TEXT_SCALE = 0.6
TEXT_THICKNESS = 1
BOX_THICKNESS = 2
```

```python
def load_config(yaml_path):
    """Loads the dataset's data.yaml file."""
    try:
        with open(yaml_path, 'r') as f:
            config = yaml.safe_load(f)
        return config['names']
    except Exception as e:
        print(f"Error loading {yaml_path}: {e}")
        print("Falling back to default class names: [Class 0, Class 1, Class 2...]")
        # Fallback in case YAML fails
        return [f"Class {i}" for i in range(3)] # Assuming 3 classes as per your prompt

def parse_ground_truth(label_path, img_shape):
    """Reads a YOLO .txt label file and converts it to absolute xyxy coordinates."""
    gts = []
    if not os.path.exists(label_path):
        return gts  # Return empty list for background images

    h, w = img_shape[:2]
    try:
        with open(label_path, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) != 5:
                    continue
                
                class_id, x_c, y_c, w_norm, h_norm = map(float, parts)
                
                # Convert from normalized xywh to absolute xyxy
                box_w = w_norm * w
                box_h = h_norm * h
                x1 = int((x_c * w) - (box_w / 2))
                y1 = int((y_c * h) - (box_h / 2))
                x2 = int(x1 + box_w)
                y2 = int(y1 + box_h)
                
                gts.append([x1, y1, x2, y2, int(class_id)])
    except Exception as e:
        print(f"Error parsing {label_path}: {e}")
    return gts

def draw_boxes(image, gts, preds, class_names):
    """Draws ground truth and prediction boxes on the image."""
    vis_image = image.copy()

    # 1. Draw Ground Truth boxes (Green)
    for gt in gts:
        x1, y1, x2, y2, class_id = gt
        label = f"GT: {class_names[class_id]}"
        
        # Draw the box
        cv2.rectangle(vis_image, (x1, y1), (x2, y2), GT_COLOR, BOX_THICKNESS)
        
        # --- MODIFICATION: Draw label INSIDE the box at the BOTTOM-RIGHT ---
        (w, h), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, TEXT_SCALE, TEXT_THICKNESS)
        # Draw the label background
        cv2.rectangle(vis_image, (x2 - w, y2 - h - 5), (x2, y2), GT_COLOR, -1)
        # Draw the label text
        cv2.putText(vis_image, label, (x2 - w + 2, y2 - 5), cv2.FONT_HERSHEY_SIMPLEX, TEXT_SCALE, (0, 0, 0), TEXT_THICKNESS, cv2.LINE_AA)
        # --- END MODIFICATION ---

    # 2. Draw Prediction boxes (Red)
    # Note: We're modifying the 'preds' object from the main script
    for box in preds.boxes:
        if box.conf[0] < CONF_THRESHOLD:
            continue
            
        x1, y1, x2, y2 = map(int, box.xyxy[0].cpu().numpy())
        conf = float(box.conf[0].cpu().numpy())
        class_id = int(box.cls[0].cpu().numpy())
        label = f"Pred: {class_names[class_id]} ({conf:.2f})"
        
        # Draw the box
        cv2.rectangle(vis_image, (x1, y1), (x2, y2), PRED_COLOR, BOX_THICKNESS)
        
        # --- MODIFICATION: Draw label INSIDE the box at the TOP-LEFT ---
        # Get text size (Corrected from previous error)
        (w, h), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, TEXT_SCALE, TEXT_THICKNESS) 
        # Draw the label background (Re-added)
        cv2.rectangle(vis_image, (x1, y1), (x1 + w, y1 + h + 5), PRED_COLOR, -1)
        # Draw the label text (Corrected)
        cv2.putText(vis_image, label, (x1 + 2, y1 + h + 2), cv2.FONT_HERSHEY_SIMPLEX, TEXT_SCALE, (255, 255, 255), TEXT_THICKNESS, cv2.LINE_AA)
        # --- END MODIFICATION ---
        
    return vis_image
```

```python

print(f"Loading model from {MODEL_PATH}...")
model = YOLO(MODEL_PATH)

print(f"Loading class names from {DATA_YAML_PATH}...")
class_names = load_config(DATA_YAML_PATH)

print(f"Finding test images in {TEST_IMG_DIR}...")
# Find all common image formats
image_patterns = [os.path.join(TEST_IMG_DIR, ext) for ext in ("*.jpg", "*.jpeg", "*.png")]
all_image_paths = []
for pattern in image_patterns:
    all_image_paths.extend(glob.glob(pattern))
    
if not all_image_paths:
    print(f"Error: No images found in {TEST_IMG_DIR}. Please check your paths.")
else:
    # Sample images
    num_to_sample = min(NUM_IMAGES, len(all_image_paths))
    print(f"Found {len(all_image_paths)} images. Randomly sampling {num_to_sample}...")
    sampled_image_paths = random.sample(all_image_paths, num_to_sample)
    
    # --- MODIFICATION: Create the grid for plotting (16 rows, 4 columns) ---
    num_cols = 4
    num_rows = int(np.ceil(num_to_sample / num_cols))
    
    fig_w = num_cols * 4  # 4 inches per image width
    fig_h = num_rows * 4  # 4 inches per image height
    
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(fig_w, fig_h))
    axes = axes.flatten() # Keep flattening to easily iterate
    # --- END MODIFICATION ---
    
    print("Running predictions and generating grid... This may take a moment.")
    
    for i, img_path in enumerate(sampled_image_paths):
        ax = axes[i]
        
        # Read image
        img_bgr = cv2.imread(img_path)
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
        
        # --- 1. Get Ground Truth ---
        label_name = os.path.basename(img_path)
        label_name = os.path.splitext(label_name)[0] + ".txt"
        label_path = os.path.join(TEST_LABEL_DIR, label_name)
        gts = parse_ground_truth(label_path, img_rgb.shape)
        
        # --- 2. Get Predictions ---
        # Run model on the BGR image (YOLO handles BGR/RGB)
        results = model(img_bgr, verbose=False, conf=CONF_THRESHOLD)
        preds = results[0] # Results for the first image
        
        # --- 3. Draw and Plot ---
        vis_image = draw_boxes(img_rgb, gts, preds, class_names)
        
        ax.imshow(vis_image)
        ax.axis('off')
        
        # --- MODIFICATION: Removed the title ---
        # ax.set_title(os.path.basename(img_path), fontsize=10) # <-- This line is removed

    # Turn off any unused axes in the grid
    for j in range(i + 1, len(axes)):
        axes[j].axis('off')
        
    plt.tight_layout()
    print("Grid generation complete. Displaying plot...")
    plt.show()
    
    # Save the figure
    output_filename = "predictions_grid.png"
    fig.savefig(output_filename)
```

```python
import os
from collections import Counter
import yaml

# ================= CONFIGURATION =================
# 1. Path to your test labels folder
# It usually looks like "BirdNest-Balanced-1/test/labels"
test_labels_path = "BirdNest-Balanced-1/test/labels"

# 2. Path to your data.yaml to get class names (optional)
# If you don't have this, set it to None, and we will just use ID numbers
yaml_path = "BirdNest-Balanced-1/data.yaml"
# =================================================

def count_instances():
    # Check if path exists
    if not os.path.exists(test_labels_path):
        print(f"Error: The path '{test_labels_path}' does not exist.")
        return

    class_counts = Counter()
    total_images = 0
    empty_images = 0

    # Get all .txt files
    txt_files = [f for f in os.listdir(test_labels_path) if f.endswith('.txt')]
    total_images = len(txt_files)

    print(f"Scanning {total_images} images in '{test_labels_path}'...\n")

    for file in txt_files:
        file_path = os.path.join(test_labels_path, file)
        with open(file_path, 'r') as f:
            lines = f.readlines()
            
            if not lines:
                empty_images += 1
                continue

            for line in lines:
                # YOLO format: class_id x_center y_center width height
                # We only care about the first number (class_id)
                parts = line.strip().split()
                if parts:
                    class_id = int(parts[0])
                    class_counts[class_id] += 1

    # Try to load class names from YAML
    class_names = {}
    if yaml_path and os.path.exists(yaml_path):
        try:
            with open(yaml_path, 'r') as f:
                data = yaml.safe_load(f)
                # Handle different yaml structures (sometimes names is a list, sometimes a dict)
                if 'names' in data:
                    class_names = data['names']
        except Exception as e:
            print(f"Could not load class names: {e}")

    # --- PRINT RESULTS ---
    print(f"{'Class ID':<10} {'Class Name':<20} {'Instance Count':<15}")
    print("-" * 45)
    
    # Sort by Class ID
    for class_id in sorted(class_counts.keys()):
        # Get name if available, otherwise "Unknown"
        name = class_names[class_id] if isinstance(class_names, list) and class_id < len(class_names) else class_names.get(class_id, "Unknown")
        
        count = class_counts[class_id]
        status = "✅ Safe" if count >= 50 else "⚠️ LOW DATA" if count >= 20 else "❌ UNRELIABLE"
        
        print(f"{class_id:<10} {str(name):<20} {count:<15} {status}")

    print("-" * 45)
    print(f"Total Objects: {sum(class_counts.values())}")
    print(f"Empty Images: {empty_images}")

if __name__ == "__main__":
    count_instances()
```
